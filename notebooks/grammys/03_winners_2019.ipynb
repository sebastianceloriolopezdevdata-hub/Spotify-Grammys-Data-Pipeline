```python
import pandas as pd
```

```python
# Load Grammy winners data
df_winners = pd.read_csv("../../data/raw/grammys_2019.csv")

# Structural Overview
print(f"Grammy Winners Dataset Shape: {df_winners.shape}")
print("-" * 50)
df_winners.info()
```

```python
# Check the extracted data
print("Total categories:", df_winners["Category"].nunique())
print("Missing categories:", df_winners["Category"].isna().sum())

display(df_winners.tail(10))
```

```python
print(f"Dataset shape: {df_winners.shape}")
print(f"Unique categories: {df_winners['Category'].nunique()}")

print("\nMissing values:")
print(df_winners.isna().sum())
```

```python
# Check for exact duplicate records
exact_duplicates = df_winners[
    df_winners.duplicated(keep=False)
]

print(f"Rows involved in exact duplicates: {len(exact_duplicates)}")

display(exact_duplicates)
```

```python
# Count records per category
category_counts = (
    df_winners["Category"]
    .value_counts()
    .reset_index()
)

category_counts.columns = ["Category", "Frequency"]

display(
    category_counts[
        category_counts["Frequency"] > 1
    ]
)
```
